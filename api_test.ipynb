```python
import requests

url = "https://api.maxar.com/discovery/v1/search"

params = {
    "ids": "1020010079424900"  # ID yang pasti ada
}

headers = {
  'Accept': 'application/geo+json',
  'MAXAR-API-KEY': 'Yzk1ZmM1ZGItOTBiMi00ZjUyLWE5ZjItY2MxMDYwZDFmZDJi'
}

response = requests.get(url, headers=headers, params=params)

if response.status_code == 200:
    data = response.json()
    features = data.get("features", [])
    print("Number of features:", len(features))
    for item in features:
        props = item.get("properties", {})
        print("ID:", item.get("id"))
        print("Datetime:", props.get("datetime"))
        print("Collection:", props.get("collection"))
        print("---")
else:
    print("Error:", response.status_code, response.text)
```
